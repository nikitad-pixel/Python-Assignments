```python
#3 program to check whether a figure is right angled triangle
def right_angled_triangle():
 a=float(input("length of a:"))
 b=float(input("length of b:"))
 c=float(input("length of c:"))

 sides=[a,b,c]
 sides.sort()

#using pythagoras theorem
 if sides[0]**2+sides[1]**2==sides[2]**2:
     print("this is a right angled triangle")
 else:
    print("this is not a right angled triangle")

right_angled_triangle()
```
