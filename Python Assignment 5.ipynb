```python
#PAN Number Validation Using Regular Expression
import re
def isvalidPanCardno(panCardNo):
    regex="[A-Z]{5}[0-9]{4}[A-Z]{1}"

    p=re.compile(regex)

    if (panCardNo==None):
        return False

    if(re.search(p, panCardNo)and len(panCardNo)==10):
        return True
    else:
        return False

#test 1:
str1="BNZAA2318J"
print(isvalidPanCardno(str1))

# test 2:
str2 = "23ZAABN18J"
print(isvalidPanCardno(str2))

# test 3:
str3 = "BNZAA2318JM"
print(isvalidPanCardno(str3))

# test 4:
str4 = "BNZAA23184"
print(isvalidPanCardno(str4))

# test 5:
str5 = "BNZAA 23184"
print(isvalidPanCardno(str5))

```
